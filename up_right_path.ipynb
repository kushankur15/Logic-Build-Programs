```python
# defining the code for route from (0,0) to (n,n)
def routes(n):
    start_point = (0,0)
    end_point = (n,n)

    points = []
    for i in range(n+1):
        for j in range(i, n+1):
            points.append((j,i))

    edge_points = []
    for point in points:
        if point[0] == point[1] or point[0] == n:
            edge_points.append(point)

    paths = []
    for i in range(1000):
        curr_point = (0, 0)
        next_point = (0, 0)
        route=[]
        while(curr_point!=(n,n)):
            if curr_point in edge_points:
                if curr_point[0] == n:
                    next_point = (curr_point[0], curr_point[1] + 1)
                else:
                    next_point = (curr_point[0] + 1, curr_point[1])
            else:
                next_points = [(curr_point[0] + 1, curr_point[1]), (curr_point[0], curr_point[1] + 1)]
                next_point = random.choice(next_points)
            curr_point = next_point
            route.append(next_point)
        paths.append(route)
    return points,paths
```
